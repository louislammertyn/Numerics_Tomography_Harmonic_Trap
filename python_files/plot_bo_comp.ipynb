```python
import numpy as np

data = np.load("results_bo_comparison.npz")

num_ex = data["num_ex"]
num_eff = data["num_eff"]
exp = data["exp"]


# Numerical solutions
t_ex = num_ex[:, 0]
y_ex = num_ex[:, 1]

t_eff = num_eff[:, 0]
y_eff = num_eff[:, 1]

# Experimental
t_exp = exp[:, 0]
y_exp = exp[:, 1]
y_std = exp[:, 2]
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))

# Numerical solutions
plt.plot(t_ex, y_ex, label="Numerical (exact)", linewidth=2)
plt.plot(t_eff, y_eff, "--", label="Numerical (effective)", linewidth=2)

# Experimental with error bars
plt.errorbar(
    t_exp, y_exp, yerr=y_std,
    fmt='o--', markersize=5,
    capsize=3,
    label="Experimental",
    color="black"
)

# Labels & styling
plt.xlabel("t (ms)", fontsize=12)
plt.ylabel("$< \lambda_{com} >$", fontsize=12)
plt.legend()
plt.grid(alpha=0.3)

# Clean professional look
plt.tight_layout()

# Save high-quality figure
plt.savefig("bo_comparison.png", dpi=300)
plt.show()
```
